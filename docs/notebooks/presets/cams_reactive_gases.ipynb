# CAMS reactive & pollutant gases

The **reactive/pollutant gas** presets from the CAMS composition set: surface **ozone** and **nitrogen dioxide**, plus the **total-column** ozone, carbon monoxide and sulphur dioxide fields.\n\nEach renders with its reference discrete-band colour scale over the parameter's own fixed value range. Pass any raster of the matching quantity under the preset's key to get the same authentic look.

## Setup

Import the one-call styler and the preset registry, and build a small synthetic-field helper.

```python
import numpy as np
import matplotlib.pyplot as plt

from cleopatra.styling.colors import apply_data_style, DATA_STYLES


def demo_field(name, shape=(90, 140)):
    """A smooth synthetic field whose values span the preset's own colour bands.

    We build a couple of Gaussian bumps, normalise to 0..1, then map that
    through the preset's `levels` so every discrete band actually shows.
    """
    layer = DATA_STYLES[name][name]
    levels = layer.get("levels")
    ny, nx = shape
    yy, xx = np.mgrid[0:ny, 0:nx]

    def blob(cx, cy, sx, sy):
        return np.exp(-(((xx - cx) / sx) ** 2 + ((yy - cy) / sy) ** 2))

    field = 0.25 * blob(40, 28, 30, 20) + 0.7 * blob(96, 58, 26, 30) + 0.5 * blob(70, 74, 42, 16)
    field = (field - field.min()) / (field.max() - field.min())
    if levels is None:
        return field
    return np.interp(field, np.linspace(0.0, 1.0, len(levels)), levels)


def gallery(names, ncols=3):
    """Render each named preset on its own demo field in a grid (no legend)."""
    nrows = int(np.ceil(len(names) / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=(4.0 * ncols, 3.0 * nrows))
    axes = np.atleast_1d(axes).ravel()
    for ax, name in zip(axes, names):
        apply_data_style(ax, {name: demo_field(name)}, style=name, legend=False)
        label = DATA_STYLES[name][name]["label"]
        ax.set_title(f"{name}\n{label}", loc="left", fontsize=8)
        ax.set_xticks([])
        ax.set_yticks([])
    for ax in axes[len(names):]:
        ax.set_visible(False)
    fig.tight_layout()
    plt.show()
```

## The preset gallery

Every preset in this group on its own demo field.

```python
gallery(['ozone', 'nitrogen_dioxide', 'total_column_ozone', 'total_column_carbon_monoxide', 'total_column_sulphur_dioxide'])
```

## One preset in detail

The full styled output, this time with the labelled band legend `apply_data_style` draws by default.

```python
name = "total_column_ozone"
label = DATA_STYLES[name][name]["label"]
fig, ax = plt.subplots(figsize=(7.5, 4.5))
apply_data_style(ax, {name: demo_field(name)}, style=name)
ax.set_title(f'style="{name}"  ({label})', loc="left")
ax.set_xticks([])
ax.set_yticks([])
plt.show()
```
