# Operational weather parameter presets

The operational long tail of the vendored weather library: marine/wave fields, upper-air temperature / wind / geopotential, dynamics (vorticity, potential vorticity, divergence, vertical velocity), cloud and humidity, hydrology, and instability indices. Each carries the parameter's reference banded look; the demo fields are synthetic (cleopatra never reads files -- NumPy in, matplotlib out).

## Setup

```python
import numpy as np
import matplotlib.pyplot as plt

from cleopatra.styling.colors import DATA_STYLES, apply_data_style


def demo_field(name, shape=(80, 120)):
    """A smooth synthetic field spanning the preset's own scale (levels or vmin/vmax; 0..1 for a plain colour map)."""
    layer = DATA_STYLES[name][name]
    ny, nx = shape
    yy, xx = np.mgrid[0:ny, 0:nx]

    def blob(cx, cy, sx, sy):
        return np.exp(-(((xx - cx) / sx) ** 2 + ((yy - cy) / sy) ** 2))

    field = 0.25 * blob(38, 26, 30, 20) + 0.7 * blob(92, 55, 26, 30) + 0.5 * blob(66, 70, 40, 16)
    field = (field - field.min()) / (field.max() - field.min())
    levels = layer.get("levels")
    if levels:
        return np.interp(field, np.linspace(0.0, 1.0, len(levels)), levels)
    vmin, vmax = layer.get("vmin"), layer.get("vmax")
    if vmin is not None and vmax is not None:
        return vmin + field * (vmax - vmin)
    return field


def gallery(names, ncols=4):
    """Render each named preset on its own demo field in a grid (no legend)."""
    nrows = int(np.ceil(len(names) / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=(3.3 * ncols, 2.5 * nrows))
    axes = np.atleast_1d(axes).ravel()
    for ax, name in zip(axes, names):
        apply_data_style(ax, {name: demo_field(name)}, style=name, legend=False)
        ax.set_title(name, loc="left", fontsize=8)
        ax.set_xticks([])
        ax.set_yticks([])
    for ax in axes[len(names):]:
        ax.set_visible(False)
    fig.tight_layout()
    plt.show()
```
